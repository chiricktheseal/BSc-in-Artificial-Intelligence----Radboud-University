#### Question 4:

$p(C=H \mid D)$ does not sum to 1 because C=T is not considered. 
$p(C=H \mid D)$ is a conditional probability, not a marginal probability. 
It does not sum to 1 because it represents probabilities conditioned on different values D, but not the probabilities over all possible outcomes of C or D!
It is like considering only one of two branches in a tree. From the node D there are two possible events that can happen (C=H or C=T) and the probability of both of them together would equal 1.

# Calculus refresher

In practice, the situations we are interested in often concern many variables, which means their joint probability tables become too high-dimensional to write down. At the same time, we often encounter continuous variables rather than discrete ones. That means they can take an infinite number of different values, which again we cannot write down in a probability table. Instead, we have to use some calculus to work with such distributions. We will do so many times throughout this course, but for now we will simply look at finding the maximum of a probability density function. This corresponds to the most likely value for our variable!

## The value that maximizes the probability density function

An example of a probability density function is the following:

$$ p(\theta \mid a, b) = \frac{\theta^{a-1} (1-\theta)^{b-1}}{B(a,b)}\qquad 0 \leq \theta \leq 1\enspace. $$

This is a probability distribution over the variable $\theta$. The distribution has two parameters, $a$ and $b$. The function $B(a,b)$ you may ignore for now; it simply ensures that the distribution integrates to 1, as it should (we will discuss this distribution at length in lecture 2). Here are some examples of what this distribution may look like, depending on how we select $a$ and $b$:

```python
import matplotlib.pyplot as plt
plt.rc('font', size=14)         # default fontsize
plt.rc('axes', titlesize=16)    # fontsize of the axes title
plt.rc('axes', labelsize=16)    # fontsize of the x and y labels
plt.rc('xtick', labelsize=14)   # fontsize of the tick labels
plt.rc('ytick', labelsize=14)   # fontsize of the tick labels
plt.rc('legend', fontsize=16)   # legend fontsize
plt.rc('figure', titlesize=16)  # fontsize of the figure title
plt.rc('figure', titlesize=16)  # fontsize of the figure title


import numpy as np
from scipy.special import beta

def probability_density_function(theta, a, b):
    return theta**(a-1) * (1-theta)**(b-1) / beta(a,b)

theta = np.linspace(0, 1, num=100)

a_ = [3, 7, 2]
b_ = [2, 9, 5]

plt.figure()
for a, b in zip(a_, b_):
    plt.plot(theta, probability_density_function(theta, a, b), lw=2, label=f'$a={a}, b={b}$')
plt.xlim([0, 1])
plt.xlabel(r'$\theta$')
plt.ylabel(r'$p(\theta \mid a, b)$')
plt.ylim(bottom=0)
plt.legend()
plt.show()
```

Recall, to find the maximum of such a function, first take its logarithm, discard all terms that do not depend on $\theta$, find the derivative and set it to zero.

### Answer: The value that maximizes the probability density function

#### Question 1:

$$
\begin{align*}
\hat{\theta} &= \underset{\theta}{\operatorname{argmax}} p(\theta\mid a, b) \enspace \\
&= \underset{\theta}{\operatorname{argmax}} \frac{\theta^{a-1} (1-\theta)^{b-1}}{B(a,b)}\qquad 0 \leq \theta \leq 1\enspace\\
\\
\\
\log  p(\theta\mid a, b) &= \log (\frac{\theta^{a-1} (1-\theta)^{b-1}}{B(a,b)})\\
&= \log (\theta^{a-1}(1-\theta)^{b-1}) - \log (B(a,b))\\
&= \log(\theta^{a-1}) + \log((1-\theta)^{b-1}) - \log (B(a,b))\\
&= (a-1)\log(\theta) + (b-1) \log(1-\theta) - \log (B(a,b))\\
\\
\\
\frac{\partial \log p(\theta\mid a, b)}{\partial \theta} &= \frac{\partial}{\partial \theta}((a-1)\log(\theta) + (b-1) \log(1-\theta) - \log (B(a,b))) \\
&= \frac{\partial}{\partial \theta}((a-1)\log(\theta)) + \frac{\partial}{\partial \theta}((b-1) \log(1-\theta)) - \frac{\partial}{\partial \theta}\log (B(a,b))\\
&= (a-1)\frac{\partial}{\partial \theta}\log(\theta) + (b-1)\frac{\partial}{\partial \theta}\log(1-\theta)\\
&= (a-1)\frac{1}{\theta} + (b-1) \frac{1}{\theta - 1}\\
\\
\\
0 &= (a-1)\frac{1}{\theta} + (b-1) \frac{1}{\theta - 1}\\
&= (a-1)\frac{\theta - 1}{\theta(\theta-1)} + (b-1)\frac{\theta}{\theta(\theta-1)}\\
&= \frac{(a-1)(\theta - 1) + (b-1)\theta}{\theta(\theta-1)}\\
&= \frac{a\theta - a - \theta + 1 + b\theta - \theta}{\theta(\theta-1)}\\
&= a\theta - a - \theta + 1 + b\theta - \theta \\
&= \theta(a - 2 + b) - a + 1 \\
a - 1 &= \theta(a + b - 2)\\
\frac{a - 1}{a + b - 2} &= \theta
\end{align*}
$$
The final formula is $\hat{\theta} = \dfrac{a - 1}{a + b - 2}$

Therefore, the maxima of the three functions are:
$$
\begin{align*}
\text{For a = 3, b = 2: }\hspace{1cm} \hat{\theta} &= \dfrac{3 - 1}{3 + 2 - 2} = \frac{2}{3}\\
\\
\text{For a = 7, b = 9: }\hspace{1cm} \hat{\theta} &= \dfrac{7 - 1}{7 + 9 - 2} = \frac{6}{14} = \frac{3}{7}\\
\\
\text{For a = 2, b = 5: }\hspace{1cm} \hat{\theta} &= \dfrac{2 - 1}{2 + 5 - 2} = \frac{1}{5}
\end{align*}
$$

#### Question 2:

Compute the integral:

$$
\begin{align*}
\int_0^1 p(x) dx &= \int_0^1 6x(1-x) dx \\
&= 6 \int_0^1 x(1-x) dx \\
&= 6 \int_0^1 (x-x^2) dx \\
&= 6 (\int_0^1 x dx - \int_0^1 x^2 dx) \\
&= 6 ([\frac{1}{2}x^2]_0^1 - [\frac{1}{3}x^3]_0^1) \\
&= 6 (\frac{1}{2}1^2 - \frac{1}{2}0^2 - (\frac{1}{3}1^3 - \frac{1}{3}0^3))\\
&= 6 (\frac{1}{2} - \frac{1}{3})\\
&= 6 \cdot \frac{1}{6}\\
&= 1
\end{align*}
$$
As computed, the integral is indeed 1 and as the plot shows, the function p(x) is positive for all values between 0 and 1.
