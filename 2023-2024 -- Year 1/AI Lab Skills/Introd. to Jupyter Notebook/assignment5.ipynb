### 2.1 Matrix product by hand
Let's assume we have the following two matrices:

$$\mathbf{A} = \begin{bmatrix}2 & 1 & -4\\3 & -2 & 1\end{bmatrix}$$

$$\mathbf{B} = \begin{bmatrix}-1 & 2 &\\2 & -2 \\1 & 3\end{bmatrix}$$

Compute -- with pen and paper -- the matrix product $\mathbf{C}=\mathbf{A}\mathbf{B}$. Write down, in LaTeX, all steps (i.e., not just the end results) required to compute this product.

# Answer here

Latex Code:

#required \usepackage{amsmath}  
$
AB = \begin{bmatrix} 2 & 1 & -4 \\ 3 & -2 & 1 \end{bmatrix} \cdot 
        \begin{bmatrix} 1 & -2 \\ 2 & -2 \\ 1 & 3 \end{bmatrix}
$  

$
AB_{11} = (2 \cdot -1) + (1 \cdot 2) + (-4 \cdot 1) = (-2) + 2 - 4 = -4
$  
$
AB_{12} = (2 \cdot 2) + (1 \cdot -2) + (-4 \cdot 3) = 44 - 2 - 12 = -10

$

$

### 2.3 Matrix product with for loops
Compute $\mathbf{C}=\mathbf{A}\mathbf{B}$ with for loops and without any built-in Python function or any NumPy function. Print the result and compare to your own computation from 1.1.

```python
# Answer here

# result is 2x2
result = [[0,0],[0,0]]

# iterate through rows of A
for i in range(len(A)):
    # iterate through columns of B
    for j in range(len(B[0])):
        # iterate through rows of B
        for k in range(len(B)):
            result[i][j] += A[i][k] * B[k][j]

for r in result:
      print(r)
```

### 3.1 Represent the grades and the weights as NumPy arrays

```python
# Answer

# Grades for Anne, Bob, and Chris
ABC_grades = np.array([
[10.0, 8.0, 3.0, 7.0, 8.0, 8.0, 9.0],
[10.0, 9.0, 8.5, 4.0, 9.0, 8.0, 6.0],
[6.5, 7.0, 8.0, 6.0, 8.0, 7.5, 10.0]
])

# Weights for assignments
weights_data = np.array([1/12, 1/12, 1/12, 1/12, 1/12, 1/12, 6/12])

# Print the arrays and their dimensions
print("shape grades: ", ABC_grades.shape )
print("shape weights: ", weights_data.shape )

# Print the grades matrix and weights vector
print("Grades: " )

students = ['Anne', 'Bob', 'Chris']
assignments = ['A1', 'A2', 'A3', 'A4', 'A5', 'A6', 'E1']

for i, student in enumerate(students):
    print(f"    {student}:".ljust(12), end="")
    for j, assignment in enumerate(assignments):
        print(f"  {assignment}:",f"{ABC_grades[i, j]:.1f}".rjust(4), end="")
    print()
```

```python
# Answer here

# 3.2.1 Compute the (unweighted) average grade for each student
average_grades = np.mean(ABC_grades, axis = 1)
print("Average grades per student: ",average_grades)

# 3.2.2 Compute the average grade per assignment (excluding the exam)
average_grades_assignments = np.mean(ABC_grades[:, :-1], axis=0)
print("Average grades per assignment (excluding exam): ", average_grades_assignments)

# 3.2.3 Compute the (weighted) final grades for each student
final_grades = np.dot(ABC_grades, weights_data)
print("Final grades for each student: ", final_grades)
```

## 4. Eigenvalue decomposition
Let's assume we have some two-dimensional data that follows a multivariate Gaussian distribution. As an analogy of this dataset, think of a swarm of bees that are flying at a particular 2D coordinate ($x$, $y$). As the swarm follows a Gaussian distribution, it simply means they typically like to be close to one another.

We set the mean of the 2D Gaussian to zero, while having some covariance (i.e., the two variables $x$ and $y$ interact). Inspect the code below, which uses the `np.random.multivariate_normal()` function to draw random samples of the 2D Gaussian that we define with `mu` (the means) and `cov` (the covariance). 

Note, we input `n_samples`, telling the function we want to draw `n_samples` observations (e.g., the number of bees). 

Note, we first called the `np.random.seed()` function to set the random seed of the random number generator. This ensures that every call to this pseudo-random number generator will give the same "random" examples, so that all the assignments use the same samples.

```python
# Generate some random Gaussian distributed data
np.random.seed(0)
mu = np.array([0, 0])  # mean
cov = np.array([[5, 3], [3, 4]])  # covariance
n_samples = 1000  # number of observations
data = np.random.multivariate_normal(mu, cov, n_samples)

# Print shapes
print("shape mu:", mu.shape)
print("shape cov:", cov.shape)
print("shape data:", data.shape)
```

### 4.1 Visualize the data
Visualize the data in a scatterplot, which shows a dot for every data-point. Feel free to continue using the bee example, where every data-point is a bee which is at a particular ($x$, $y$) location. Again, make sure to annotate your figure as much as possible (i.e., add labels etc.).

Additionally, plot the covariance matrix as an image in a separate figure. Add a colorbar to the image. In this image you should be able to see the variance and covariance of the variables.

```python
# Answer here

# Data
x = data[:, 0] #extracts all values of the first column from the data array and assigns them to the variable x
y = data[:, 1]

# Create a new figure
plt.figure(figsize=(8, 6))

# Scatterplot
plt.scatter(x, y, marker='o')

# Add labels and legend
plt.xlabel('Variable X')
plt.ylabel('Variable Y')
plt.title('Scatter Plot of 2D Gaussian Dat')

#plt.show()


# Plot the covariance matrix
plt.figure(figsize=(8, 6))
plt.imshow(cov, cmap='viridis')  # отображение матрицы
plt.colorbar()  # add a colorbar to the image

# Add labels and legend
plt.xticks(range(len(mu)), ['X', 'Y'])
plt.yticks(range(len(mu)), ['X', 'Y'])
plt.title('Covariance Matrix')

#plt.show()
```

### 4.2 Eigenvalue decomposition
You should see that the data are rotated. This is also visible in the covariance matrix, because it shows a covariance of 3 between $x$ and $y$. In many domains, such as machine learning, you want to correct for this. Specifically, you want to make your variables (i.e., your features) in your data orthogonal (so that they do not interact). 

Perform the eigenvalue decomposition on the covariance matrix using `np.linalg.eig()`. This function gives you the eigenvalues and eigenvectors. Then, print the shapes of these two variables. Subsequently, visualize the scatterplot of your data like you did before. In the scatterplot, draw the eigenvectors as arrows in the plot using `plt.arrow()`. Draw the first eigenvector in cyan color and the second component in magenta color. To get the best visualization, call `plt.axis('square')` to make sure that the x and y axis are of equal limits. In this way, you should see that the eigenvectors are orthogonal, which is a property of the eigenvalue decomposition. 

**Note**, the eigenvectors start from the origin and have a direction as defined by the eigenvector and a length as defined by the product between the eigenvector and the quare-root of its eigenvalue. 

**Hint:** because our data is 2D (i.e., we have two variables, $x$ and $y$), we will have two components, so two eigenvalues each with two values (one for each dimension) and two eigenvectors. Thus, the eigenvector variable is a $2 \times 2$ matrix. Make sure to index the correct dimension to extract eigenvectors (look at the documentation of `np.linalg.eig()`). 

```python
# Answer here

# Eigenvalue decomposition of the covariance matrix
eigenvalues, eigenvectors = np.linalg.eig(cov)

# Print the shapes of eigenvalues and eigenvectors
print("Eigenvalues shape: ", eigenvalues.shape)
print("Eigenvectors shape: ", eigenvectors.shape)


# Create a new figure
plt.figure(figsize=(8, 6))

# Scatterplot
plt.scatter(x, y, marker='o', label='Data')

# Draw the second eigenvector in magenta color
plt.arrow (0, 0, eigenvectors[0, 0] * np.sqrt(eigenvalues[0]), eigenvectors[1, 0] * np.sqrt(eigenvalues[0]), width=0.1, head_width=0.2, color='cyan', label='Eigenvector 1')
# Draw the first eigenvector in cyan color
plt.arrow(0, 0, eigenvectors[0, 1] * np.sqrt(eigenvalues[1]), eigenvectors[1, 1] * np.sqrt(eigenvalues[1]),  width=0.1, head_width=0.2, color='magenta', label='Eigenvector 2')

# Set equal limits for x and y axes
plt.axis('square') 

# Add labels and legend
plt.xlabel('Variable X')
plt.ylabel('Variable Y')
plt.title('Scatter Plot with Eigenvectors')
plt.legend()

#plt.show()
```

### 4.3 Rotating the data
Now that you have the eigenvalue decomposition, rotate the data so that the variables $x$ and $y$ become orthogonal (in other words, the eigenvectors become the new set of basis vectors). Do so by computing the matrix product between the data and the eigenvectors. Print the dimensions of the rotated data; they should be equal to those of the original data (i.e., we still have an equal number of dimensions and observations). 

Subsequently, compute the covariance matrix of the rotated data using the `np.cov()` function. Print the dimensions of the new covariance matrix; they should be equal to those of the original data (i.e., we still have an equal number of dimensions). 

Furthermore, compute the eigenvalue decomposition of the rotated data. Again, print their dimensionality and compare to the original ones to verify correctness.

Now, create a figure in which we can visualize two axes, because we want to visualize two side-by-side plots. In the left plot, visualize the same as in the previous exercise (i.e., the scatter plot and the eigenvectors of the original data). In the right plot, make a similar scatter plot but now with the rotated data and its eigenvectors. 

You should see that the original data (left plot) was rotated (like we saw before), but that with the eigenvalue decomposition we can undo this rotation, so that the variables become orthogonal and the new data (right plot) "unrotated" (i.e., they point in the direction of the main axes (x and y)). 

Finally, plot the covariance matrix of the new data in a separate figure as an image. Add a colorbar to the image. In this image you should be able to see the variance and covariance of the variables. You should now see that there is no covariance anymore, and should be able to see that the first dimension has larger variance than the second dimension.

**FYI:** the eigenvalue decomposition is one way to perform a principal component analysis (PCA), which is a method that extracts linear subspaces from high-dimensional data with largest variance. In our example here, we rotated the data such that the first dimension contains the largest variance, the second less, etc.

```python
# Answer here

# Rotate the data
rotate_data = np.dot(data, eigenvectors)
# Print dimensions of rotated data
print("Dimensions of rotated data: ", rotate_data.shape)

# Compute the covariance matrix of the rotated data
cov_matrix = np.cov(rotate_data, rowvar=False)
# Print dimensions of the new covariance matrix
print("Dimensions of the new covariance matrix: ", cov_matrix.shape)

# Eigenvalue decomposition of the rotated covariance data
eigenvalues_rotated_data, eigenvectors_rotated_data = np.linalg.eig(cov_matrix)
# Print dimensions of the rotated eigenvalues and eigenvectors
print("Dimensions of rotated eigenvalues: ", eigenvalues_rotated_data.shape)
print("Dimensions of rotated eigenvectors: ", eigenvectors_rotated_data.shape)


# Create a figure with two subplots
fig, axs = plt.subplots(1, 2, figsize=(24, 12))


# Scatterplot
axs[0].scatter(x, y, marker='o', label="Data")

# Draw the second eigenvector in magenta color
axs[0].arrow (0, 0, eigenvectors[0, 0] * np.sqrt(eigenvalues[0]), eigenvectors[1, 0] * np.sqrt(eigenvalues[0]), width=0.1, head_width=0.2, color='cyan', label='Eigenvector 1')
# Draw the first eigenvector in cyan color
axs[0].arrow(0, 0, eigenvectors[0, 1] * np.sqrt(eigenvalues[1]), eigenvectors[1, 1] * np.sqrt(eigenvalues[1]),  width=0.1, head_width=0.2, color='magenta', label='Eigenvector 2')

# Add labels and legend
axs[0].set_xlabel('Variable X')
axs[0].set_ylabel('Variable Y')
axs[0].set_title('Original Data and Eigenvectors')
axs[0].legend()


# Scatterplot
axs[1].scatter(rotate_data[:, 0], rotate_data[:, 1], marker='o',  label="Rotated Data")

# Draw the second eigenvector in magenta color
axs[1].arrow (0, 0, eigenvectors_rotated_data[0, 0] * np.sqrt(eigenvalues_rotated_data[0]), eigenvectors_rotated_data[1, 0] * np.sqrt(eigenvalues_rotated_data[0]), width=0.1, head_width=0.2, color='cyan', label='Rotated Eigenvector 1')
# Draw the first eigenvector in cyan color
axs[1].arrow(0, 0, eigenvectors_rotated_data[0, 1] * np.sqrt(eigenvalues_rotated_data[1]), eigenvectors_rotated_data[1, 1] * np.sqrt(eigenvalues_rotated_data[1]),  width=0.1, head_width=0.2, color='magenta', label='Rotated Eigenvector 2')

# Add labels and legend
axs[1].set_xlabel('Variable X')
axs[1].set_ylabel('Variable Y')
axs[1].set_title('Scatter Plot with Eigenvectors')
axs[1].legend()

#plt.show()


# Plot the covariance matrix of the rotated data
plt.figure(figsize=(8, 6))

plt.imshow(cov_matrix, cmap='viridis')

plt.colorbar()  # add a colorbar to the image

# Add labels and legend
plt.xticks(range(len(mu)), ['X (Rotated)', 'Y (Rotated)'])
plt.yticks(range(len(mu)), ['X (Rotated)', 'Y (Rotated)'])
plt.title('Covariance Matrix of Rotated Data')

#plt.show()
```
